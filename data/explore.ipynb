```python
import pandas as pd

# load whatever CSVs were generated
wrote = pd.read_csv("triples_wrote.csv")
influenced = pd.read_csv("triples_influenced_by.csv")
genre = pd.read_csv("triples_has_genre.csv")
topics = pd.read_csv("triples_has_topic.csv")
books = pd.read_csv("entities_books.csv")
authors = pd.read_csv("entities_authors.csv")

print(f"Books:           {len(books)}")
print(f"Authors:         {len(authors)}")
print(f"wrote:           {len(wrote)}")
print(f"influencedBy:    {len(influenced)}")
print(f"hasGenre:        {len(genre)}")
print(f"hasTopic:        {len(topics)}")
print(f"Books with OL ID: {books['ol_id'].notna().sum()}")
print(f"Unique themes:   {topics['object'].nunique()}")
```

```python
# how many books share at least 1 theme with another book?
theme_counts = topics.groupby("object")["subject"].count()
print("Themes connecting ≥2 books:", (theme_counts >= 2).sum())
print("Themes connecting ≥5 books:", (theme_counts >= 5).sum())
print("\nTop 10 most common themes:")
print(theme_counts.sort_values(ascending=False).head(10))
```
